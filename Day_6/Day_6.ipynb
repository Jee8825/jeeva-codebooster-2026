````python
def ask_llm(prompt, system_prompt="You are a helpful, concise, and knowledgeable AI assistant. Answer clearly and accurately. If you don't know something, say so honestly. Avoid unnecessary filler or repetition.", model=MODEL, max_tokens=1024, temperature=0.1):
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": prompt}
        ],
        max_tokens=max_tokens,
        temperature=temperature
    )
    return response.choices[0].message.content
````

````python
def parse_json_response(response_text):
    """
    Safely extract and parse JSON from model output.
    Handles extra text, markdown code blocks, and whitespace.
    """
    try:
        # Try direct parse first
        return json.loads(response_text.strip())
    except json.JSONDecodeError:
        pass

    # Try extracting JSON from markdown code block ```json ... ```
    match = re.search(r"```(?:json)?\s*(\{.*?\})\s*```", response_text, re.DOTALL)
    if match:
        try:
            return json.loads(match.group(1))
        except json.JSONDecodeError:
            pass

    # Try extracting raw JSON object {...}
    match = re.search(r"\{.*?\}", response_text, re.DOTALL)
    if match:
        try:
            return json.loads(match.group(0))
        except json.JSONDecodeError:
            pass

    # All attempts failed
    return {"error": "Failed to parse JSON", "raw_output": response_text}
````

````python
import json
import re

# ─────────────────────────────────────────
# JSON Parser
# ─────────────────────────────────────────

def parse_json_response(text):
    try:
        return json.loads(text.strip())
    except json.JSONDecodeError:
        pass
    match = re.search(r"```(?:json)?\s*(\{.*?\})\s*```", text, re.DOTALL)
    if match:
        try:
            return json.loads(match.group(1))
        except json.JSONDecodeError:
            pass
    match = re.search(r"\{.*?\}", text, re.DOTALL)
    if match:
        try:
            return json.loads(match.group(0))
        except json.JSONDecodeError:
            pass
    return {"error": "Could not parse JSON", "raw": text}


# ─────────────────────────────────────────
# Role + Temperature Wrapper
# ─────────────────────────────────────────

def ask_with_role_and_temp(role, question, temperature=0.7):
    system_prompt = f"You are {role}. Always respond ONLY in valid JSON format. No extra text."

    prompt = f"""Answer the following question.

Question: "{question}"

Respond with these keys:
- "role"        : the role you are playing
- "temperature" : {temperature}
- "answer"      : your response
- "tone"        : your tone (e.g. precise, creative, balanced)

JSON:"""

    raw = ask_llm(
        prompt,
        system_prompt=system_prompt,
        temperature=temperature
    )
    return raw


# ─────────────────────────────────────────
# Test 1: Same Role, Same Question, Different Temps
# ─────────────────────────────────────────

role     = "a creative writing coach"
question = "How do I overcome writer's block?"

print("=" * 60)
print(f"Role    : {role}")
print(f"Question: {question}")
print("=" * 60)

for temp in [0.0, 0.5, 1.0, 1.5]:
    print(f"\n🌡️  Temperature: {temp}")
    print("-" * 60)
    raw    = ask_with_role_and_temp(role, question, temperature=temp)
    parsed = parse_json_response(raw)
    print(f"  Tone   : {parsed.get('tone', 'N/A')}")
    print(f"  Answer : {parsed.get('answer', 'N/A')}")


# ─────────────────────────────────────────
# Test 2: Different Roles + Matched Temperatures
# ─────────────────────────────────────────

print("\n\n🎯 Best Role + Temperature Combinations")
print("=" * 60)

combos = [
    ("a data scientist",      "What is overfitting?",               0.0),
    ("a motivational coach",  "How do I build good habits?",        0.7),
    ("a poet",                "Describe what data means to you.",   1.2),
    ("a strict critic",       "Review the idea of remote work.",    0.3),
]

for role, question, temp in combos:
    print(f"\n🎭 Role : {role}  |  🌡️  Temp : {temp}")
    print(f"❓ {question}")
    print("-" * 60)
    raw    = ask_with_role_and_temp(role, question, temperature=temp)
    parsed = parse_json_response(raw)
    print(f"  Tone   : {parsed.get('tone', 'N/A')}")
    print(f"  Answer : {parsed.get('answer', 'N/A')}")
````

````python
# ── ZERO SHOT vs ONE SHOT vs ROLE SHOT ──────────────────────────────
# Using SARCASM DETECTION — a harder task where all 3 methods differ

texts = [
    "Oh great, another Monday morning meeting.",
    "Yeah because sleeping 4 hours is totally healthy.",
    "Wow, what a surprise — the train is late again.",
    "I just love waiting in long queues for hours.",
    "The weather is nice today.",
    "Sure, because I have nothing better to do.",
    "The food was okay.",
]

def clean(response):
    return response.strip().split("\n")[0].split(":")[-1].strip().split(".")[0].strip()

print(f"{'Text':<45} | {'Zero-Shot':^12} | {'One-Shot':^12} | {'Role-Shot':^12}")
print("-" * 90)

for text in texts:

    # 🔵 ZERO SHOT — No examples, no role, direct instruction
    zero = ask_llm(
        f'Is the following text Sarcastic or Not Sarcastic? Reply with one of those two labels only.\n'
        f'Text: "{text}"\nLabel:',
        temperature=0.0
    )

    # 🟡 ONE SHOT — One clear sarcasm example to guide the model
    one = ask_llm(
        f'Is the following text Sarcastic or Not Sarcastic? Reply with one of those two labels only.\n\n'
        f'Text: "Oh fantastic, another rainy day!"\nLabel: Sarcastic\n\n'
        f'Text: "{text}"\nLabel:',
        temperature=0.0
    )

    # 🟢 ROLE SHOT — Expert linguist persona
    role = ask_llm(
        f'Is the following text Sarcastic or Not Sarcastic? Reply with one of those two labels only.\n'
        f'Text: "{text}"\nLabel:',
        system_prompt=(
            "You are an expert linguist and sarcasm detection AI. "
            "You deeply understand tone, irony, and context. "
            "Reply with exactly one label: Sarcastic or Not Sarcastic."
        ),
        temperature=0.0
    )

    z = clean(zero)
    o = clean(one)
    r = clean(role)

    # Highlight where they DISAGREE
    flag = "  ⚠️ DIFF" if not (z == o == r) else ""

    print(f"{text[:44]:<45} | {z:^12} | {o:^12} | {r:^12}{flag}")

print("-" * 90)
print("⚠️  = Methods disagree on this text")
````

````python
# ══════════════════════════════════════════════════════════════
# 5 MESSY INVOICES — ONE-SHOT PROMPTING (SIMPLE)
# ══════════════════════════════════════════════════════════════

invoices = [
    ("Invoice 1 — Missing Fields", """
        INV-2024-001
        Client: Acme Corp
        Date: 3/5/24
        item: Web Design - $1500
        item: Hosting Setup - $200
        TAX: ???
        TOTAL: not calculated
        Payment: unpaid
    """),

    ("Invoice 2 — Typos & Wrong Formats", """
        Invoice No: INV2024002
        Custmer: john smith
        Dte: March fifth twenty twenty four
        Product: Laptop x2 @ $750 each
        Productt: Mouse x5 @ 25$
        Discount: ten percent
        Total: $1575 (maybe)
        Satuts: PAYED
    """),

    ("Invoice 3 — Mixed Currencies & Duplicates", """
        INVOICE #003
        Vendor: TechSupplies Ltd.
        Date: 05-03-2024
        Item: Monitor - USD 300
        Item: Keyboard - £45
        Item: Keyboard - £45
        Item: USB Hub - €25
        Subtotal: $415
        VAT: 15 percent
        Grand Total: $477.25
        Status: Pending
    """),

    ("Invoice 4 — Completely Unstructured", """
        hey this is the invoice for last month work
        i did the logo design and also some social media posts
        logo was 500 dollars and posts were like 50 each i did 8 of them
        so total should be around 900 or something
        please pay by end of month thanks
        - freelancer mike
    """),

    ("Invoice 5 — Corrupted Data", """
        ##INVOICE@@
        Inv#: INV-2O24-OO5
        Cli3nt: Gl0bal Tech!!
        D4te: 30/02/2024
        Serv1ce: Cloud Storag3 - $99/mnth x 12 = $1,18,000
        Serv1ce: API Access -- $49.99/mnth x12 = $599.88
        T0tal: $1200 approx
        St@tus: ??
    """),
]


# ══════════════════════════════════════════════════════════════
# ONE-SHOT CLEAN FUNCTION
# ══════════════════════════════════════════════════════════════

def one_shot_clean(invoice_text):
    prompt = f"""
Clean the messy invoice below.
List the issues found, then show the cleaned version in plain text.

── EXAMPLE ──────────────────────────────────────────
Messy Invoice:
  Inv: 001, Custmer: bob, Dte: jan 5 2024
  Item: Chair @ $50 x2
  Item: Chair @ $50 x2
  Total: hundrd dollars
  Status: PAYED

Issues Found:
- Typo: "Custmer" → "Customer"
- Typo: "Dte" → "Date"
- Duplicate item: Chair x2 listed twice, merged into one
- Total written in words with typo → $100.00
- Status misspelled "PAYED" → "Paid"

Cleaned Invoice:
  Invoice No  : INV-001
  Client      : Bob
  Date        : 2024-01-05
  Items       : Chair x2 @ $50.00 = $100.00
  Subtotal    : $100.00
  Tax         : N/A
  Total       : $100.00
  Status      : Paid
── NOW CLEAN THIS ───────────────────────────────────
Messy Invoice:
{invoice_text}

Issues Found:
"""
    return ask_llm(prompt, temperature=0.0)


# ══════════════════════════════════════════════════════════════
# RUN ALL 5 INVOICES
# ══════════════════════════════════════════════════════════════

for name, invoice in invoices:
    print("═" * 60)
    print(f"📄  {name}")
    print("─" * 60)
    print("🗂️  RAW:")
    print(invoice.strip())
    print("─" * 60)
    print("✅  CLEANED:")
    result = one_shot_clean(invoice)
    print(result.strip())
    print()
````
