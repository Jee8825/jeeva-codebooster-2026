```python
import pandas as pd

df = pd.read_csv('/content/Student Performance Day 1.csv')

department_counts = df['department'].value_counts().reset_index()
department_counts.columns = ['Department', 'Student Count']

total_students = len(df)
total_columns = len(df.columns)
total_departments = len(department_counts)

print(f"Total Number of Students: {total_students}")
print(f"Total Number of Columns: {total_columns}")
print(f"Total Number of Departments (Subjects): {total_departments}")
print("\nDepartment-wise Student Count:")
print(department_counts)
```

```python

highest_math_score = df['math_score'].max()
lowest_math_score = df['math_score'].min()
average_math_score = df['math_score'].mean()

print(f"Highest Math Score: {highest_math_score}")
print(f"Lowest Math Score: {lowest_math_score}")
print(f"Average Math Score: {average_math_score:.2f}")
```

```python

average_programming_score = df['programming_score'].mean()

print(f"Average Programming Score: {average_programming_score:.2f}")
```

```python

average_attendance_percentage = df['attendance_percentage'].mean()

print(f"Average Attendance Percentage: {average_attendance_percentage:.2f}%")
```

```python
gender_avg_scores = df.groupby('gender')[['math_score', 'science_score', 'english_score', 'programming_score']].mean().round(2)
print(gender_avg_scores)
```

```python
df['overall_average_score'] = df[['math_score', 'science_score', 'english_score', 'programming_score']].mean(axis=1).round(2)

print(df[['name', 'department', 'overall_average_score']].head())
```

```python
top_performers = df.sort_values(by='overall_average_score', ascending=False).head(5)
print(top_performers[['name', 'department', 'overall_average_score']])
```

```python
students_needing_attention = df.sort_values(by='overall_average_score', ascending=True).head(5)
print(students_needing_attention[['name', 'department', 'overall_average_score']])
```
