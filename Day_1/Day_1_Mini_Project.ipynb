```python
import pandas as pd
import numpy as np

# Create synthetic data to match the project requirements
data = {
    'student_id': range(1001, 1031),
    'name': ['Aarav', 'Priya', 'Rohit', 'Sneha', 'Arjun', 'Meera', 'Kiran', 'Divya', 'Rahul', 'Ananya',
             'Vikram', 'Pooja', 'Suresh', 'Kavya', 'Ajay', 'Ritu', 'Manoj', 'Swati', 'Deepak', 'Nisha',
             'Harish', 'Tanvi', 'Sanjay', 'Preeti', 'Amit', 'Rekha', 'Gaurav', 'Sunita', 'Nitin', 'Akanksha'],
    'gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female'] * 3,
    'department': ['Computer Science']*13 + ['Electronics']*6 + ['Mechanical']*6 + ['Civil']*5,
    'math_score': np.random.randint(55, 96, size=30),
    'science_score': np.random.randint(60, 95, size=30),
    'english_score': np.random.randint(55, 91, size=30),
    'programming_score': np.random.randint(40, 98, size=30),
    'attendance_percentage': np.random.randint(65, 100, size=30),
    'city': ['Mumbai', 'Delhi', 'Bangalore', 'Kolkata', 'Chennai'] * 6,
    'admission_year': [2023] * 30
}

df = pd.DataFrame(data)
# Save it to match the expected filename in your other cells
df.to_csv('/content/Student Performance Day 1.csv', index=False)
print('Dataset generated and saved as Student Performance Day 1.csv')
display(df.head())
```

```python
sorted_df = df.sort_values(by=['department', 'programming_score'], ascending=[True, False])

top_performing_students = sorted_df.groupby('department').head(3)


display(top_performing_students[['department', 'name', 'programming_score']])
```

```python
result = df.groupby('department')[[
    'math_score',
    'science_score',
    'english_score',
    'programming_score'
]].agg(['max','mean'])

print(result)
max_score = df.groupby('department')[[
    'math_score',
    'science_score',
    'english_score',
    'programming_score'
]].max()

print(max_score)
```

```python
df['average_subject_score'] = df[['math_score', 'science_score', 'english_score', 'programming_score']].mean(axis=1)
average_score_per_department = df.groupby('department')['average_subject_score'].mean()

print("Average Overall Score per Department:")
display(average_score_per_department.sort_values(ascending=False))
```
