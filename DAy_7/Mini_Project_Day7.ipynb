```python
import chromadb
import pandas as pd

# 1. Load notes from CSV
df = pd.read_csv('/content/college_notes.csv')

notes = df.iloc[:, -1].tolist()
ids = df['note_id'].astype(str).tolist()
metadatas = df[['subject', 'topic']].to_dict('records')

# 2. Setup ChromaDB
client = chromadb.Client()
collection = client.get_or_create_collection(name="smart_notes")

# Add to collection
collection.add(
    documents=notes,
    ids=ids,
    metadatas=metadatas
)

# 3. Define search queries
queries = [
    "What is neural network optimization?",
    "How to handle data in Python?",
    "Explain supervised learning.",
    "Image processing techniques.",
    "Ensemble methods in ML.",
    "How do I fix my messy and incomplete data?"
]

print("--- Search Results (Top 3 Distance Scores) ---\n")
for q in queries:
    results = collection.query(query_texts=[q], n_results=3)
    print(f"Query: {q}")
    for i in range(len(results['ids'][0])):
        print(f"  Match {i+1}: {results['documents'][0][i]} | Distance: {results['distances'][0][i]:.4f}")
    print("-" * 30)

# 4. Filtered search: Retrieve only Data Engineering notes
print("\n--- Filtered Results (Subject: Data Engineering) ---")
# Adjusted filter to match your actual CSV categories
de_results = collection.query(
    query_texts=["ETL and databases"],
    n_results=3,
    where={"subject": "Data Engineering"}
)

for i in range(len(de_results['ids'][0])):
    print(f"Note: {de_results['documents'][0][i]} | Metadata: {de_results['metadatas'][0][i]}")
```

```python
import chromadb
import pandas as pd

# 1. Using a different set of sample data for demonstration
# (Simulating a change in CSV content)
data = {
    'note_id': ['S001', 'S002', 'S003', 'M001', 'M002'],
    'subject': ['Software Engineering', 'Software Engineering', 'Software Engineering', 'Mathematics', 'Mathematics'],
    'topic': ['Design Patterns', 'Version Control', 'Agile', 'Calculus', 'Linear Algebra'],
    'content': [
        'Design patterns are typical solutions to common problems in software design.',
        'Git is a distributed version control system that tracks changes in source code.',
        'Agile is an iterative approach to software development and project management.',
        'Calculus is the mathematical study of continuous change.',
        'Linear Algebra is the branch of mathematics concerning linear equations and matrices.'
    ]
}
df = pd.DataFrame(data)

notes = df['content'].tolist()
ids = df['note_id'].tolist()
metadatas = df[['subject', 'topic']].to_dict('records')

# 2. Setup ChromaDB (Resetting for the new data)
client = chromadb.Client()
# We use a unique name for the new data demonstration
collection = client.get_or_create_collection(name="new_subject_notes")

# Add to collection
collection.add(
    documents=notes,
    ids=ids,
    metadatas=metadatas
)

# 3. Define different search queries
queries = [
    "How do I manage code changes?",
    "What are common software architecture solutions?",
    "Tell me about math involving matrices."
]

print("--- New Search Results ---\n")
for q in queries:
    results = collection.query(query_texts=[q], n_results=2)
    print(f"Query: {q}")
    for i in range(len(results['ids'][0])):
        print(f"  Match {i+1}: {results['documents'][0][i]} | Distance: {results['distances'][0][i]:.4f}")
    print("-" * 30)

# 4. Different filter: Retrieve only Software Engineering notes
print("\n--- Filtered Results (Subject: Software Engineering) ---")
sw_results = collection.query(
    query_texts=["Software development methodologies"],
    n_results=3,
    where={"subject": "Software Engineering"}
)

for i in range(len(sw_results['ids'][0])):
    print(f"Note: {sw_results['documents'][0][i]} | Topic: {sw_results['metadatas'][0][i]['topic']}")
```
