```python
dfcs=df[df['department']=='Computer Science']
print(dfcs)
print(dfcs[['name','math_score','programming_score','attendance_percentage']])
```

```python
dfcs_high_attendance = dfcs[dfcs['attendance_percentage'] > 90]
print(dfcs_high_attendance[['name','math_score','programming_score','attendance_percentage']])
```

```python
sorted_df = df.sort_values(by=['department', 'programming_score'], ascending=[True, False])

top_performing_students = sorted_df.groupby('department').head(3)


display(top_performing_students[['department', 'name', 'programming_score']])
```

```python
result = df.groupby('department')[[
    'math_score',
    'science_score',
    'english_score',
    'programming_score'
]].agg(['max','mean'])

print(result)
max_score = df.groupby('department')[[
    'math_score',
    'science_score',
    'english_score',
    'programming_score'
]].max()

print(max_score)
```

```python
df['average_subject_score'] = df[['math_score', 'science_score', 'english_score', 'programming_score']].mean(axis=1)
average_score_per_department = df.groupby('department')['average_subject_score'].mean()

print("Average Overall Score per Department:")
display(average_score_per_department.sort_values(ascending=False))
```
