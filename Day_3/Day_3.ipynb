```python
df=pd.read_csv('Sales Data Day 2.csv')
df.head()
```

```python
# 1. Check for missing values
print("Missing values per column:")
display(df.isnull().sum())

# 2. Check for duplicates
duplicate_count = df.duplicated().sum()
print(f"\nNumber of duplicate rows: {duplicate_count}")
if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicates removed.")
```

```python
# 3. Data Types Conversion
# Using dayfirst=True and format='mixed' to handle inconsistent date strings like '07-01-2024'
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True, format='mixed')

print("\nUpdated Data Types:")
display(df.dtypes)
```

```python

df['product'] = df['product'].fillna('Unknown')
df['category'] = df['category'].fillna('Unknown')

df['quantity'] = df['quantity'].fillna(df['quantity'].median())


df = df.dropna(subset=['customer_name'])

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# 4. Unique values in text columns
text_columns = ['product', 'category', 'city', 'sales_rep']
print("\nUnique values in text columns:")
for col in text_columns:
    if col in df.columns:
        unique_vals = df[col].dropna().unique()
        print(f"{col}: {unique_vals}")
```

```python

df['product'] = df['product'].fillna('Unknown')
df['category'] = df['category'].fillna('Unknown')

df['quantity'] = df['quantity'].fillna(df['quantity'].median())

df = df.dropna(subset=['customer_name'])

print("Missing values after handling:")
print(df.isnull().sum())
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True, format='mixed')

print("Date column parsed successfully.")
print(df[['order_date']].info())
```

```python
df['formatted_date'] = df['order_date'].dt.strftime('%d/%m/%Y')

display(df[['order_date', 'formatted_date']].head())
```

```python
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.month_name()

display(df[['order_date', 'year', 'month', 'month_name']].head(20))
```

```python
df['revenue']=df['quantity']*df['unit_price']
display(df[['customer_name','product','quantity','unit_price','revenue']].head(31))
print(df['revenue'].sum())
```

```python
df['day_name'] = df['order_date'].dt.day_name()

display(df[['order_date', 'day_name']].head(10))
```

```python
text_cols = ['customer_name', 'product', 'category', 'city', 'sales_rep']
for col in text_cols:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip().str.title()

def mask_name(name):
    parts = name.split()
    if len(parts) > 1:
        return f"{parts[0]} {parts[1][0]}***"
    return f"{parts[0][0]}***"

df['customer_name'] = df['customer_name'].apply(mask_name)

print("Text standardized and names masked.")
display(df[['customer_name', 'product', 'city', 'sales_rep']].head(10))
```

```python
comparison_df = pd.DataFrame()

comparison_df['Standardized Name'] = cp_df['customer_name'].dropna().astype(str).str.strip().str.title()

comparison_df['Masked Name'] = comparison_df['Standardized Name'].apply(mask_name)

display(comparison_df.head(10))
```

```python
print("--- Missing Values Check ---")
print(df.isnull().sum())

print("\n--- Masking Validation (Sample) ---")
display(df[['customer_name']].head())

print("\n--- New Columns Validation ---")
new_cols = ['year', 'month_name', 'day_name', 'revenue']
display(df[new_cols].describe(include='all'))

clean = (df.isnull().sum() == 0).all() and (df.duplicated().sum() == 0)
print(f"\nData is clean: {clean}")

print(f"\nFinal Row Count: {len(df)}")
print(f"Total Revenue across all records: ‡{df['revenue'].sum():,.2f}")
```
